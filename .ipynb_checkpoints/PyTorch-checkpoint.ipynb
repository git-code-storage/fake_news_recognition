### Шаг 2. Подготовка данных и токенизация

```python
corpus = []
labels = []
for index, row in df.iterrows():
    document = row['title'] + ' ' + row['news_url'] + ' ' + row['source_domain']
    corpus.append(document)
    labels.append(row['real'])
print(f'Размер корпуса {len(corpus)}')
print(f'Размер списка индексов {len(labels)}')
```

```python
class_indices = {}
for i, lbl in enumerate(labels):
    if lbl not in class_indices:
        class_indices[lbl] = []
    class_indices[lbl].append(i)
for key in class_indices.keys():
    print(f'Класс {key} содержит {len(class_indices[key])} документов')
```

```python
train_ratio = 0.7
val_ratio = 0.15
test_ratio = 1 - train_ratio - val_ratio
```

```python
train_indices = []
val_indices = []
test_indices = []

for lbl, idxs in class_indices.items():
    n = len(idxs)
    n_train = int(n * train_ratio)
    n_val = int(n * val_ratio)
    
    # перемешиваем внутри класса, чтобы не было смещения по порядку
    np.random.shuffle(idxs)
    
    train_indices.extend(idxs[:n_train])
    val_indices.extend(idxs[n_train:n_train+n_val])
    test_indices.extend(idxs[n_train+n_val:])
```

```python
np.random.shuffle(train_indices)
np.random.shuffle(val_indices)
np.random.shuffle(test_indices)
```

```python
test_ind = []
for index, document in enumerate(corpus):
    if index in train_indices:
        test_ind.append(index)
print (len(test_ind))
```

### Шаг 5. Dataset + collate_fn + DataLoader

```python
class TextDataset(Dataset):
    def __init__(self, texts, labels, word2idx):
        self.texts = texts
        self.labels = labels
        self.word2idx = word2idx

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        tokens = simple_preprocess(self.texts[idx])
        ids = [self.word2idx.get(t, 1) for t in tokens]  # 1 = <UNK>
        return ids, self.labels[idx]
```

```python
def collate_fn(batch):
    sequences, labels = zip(*batch)
    padded = pad_sequence(
        [torch.tensor(s, dtype=torch.long) for s in sequences],
        batch_first=True,
        padding_value=0  # 0 = <PAD>
    )
    lengths = torch.tensor([len(s) for s in sequences], dtype=torch.long)
    labels = torch.tensor(labels, dtype=torch.float32).unsqueeze(1)
    return padded, labels, lengths
```

```python
dataset = TextDataset(corpus, labels, word2idx)
```

```python
train_subset = Subset(dataset, train_indices)
val_subset = Subset(dataset, val_indices)
test_subset = Subset(dataset, test_indices)
```

```python
train_dataloader = DataLoader(
    train_subset,
    batch_size=BATCH_SIZE,
    shuffle=True,          
    num_workers=4,
    collate_fn=collate_fn,
    drop_last=True
)
```

```python
val_dataloader = DataLoader(
    val_subset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    collate_fn=collate_fn,
    drop_last=False
)
```

```python
test_dataloader = DataLoader(
    test_subset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    collate_fn=collate_fn,
    drop_last=False
)
```

### Шаг 5. PyTorch

```python
pos_weight = torch.tensor([class_weights[1] / class_weights[0]])
pos_weight
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
```

```python
criterion = nn.BCEWithLogitsLoss(pos_weight=pos_weight)
criterion = criterion.to(device)
```

```python
def binary_accuracy(preds, y):
    rounded_preds = torch.round(torch.sigmoid(preds))
    correct = (rounded_preds == y).float()
    acc = correct.sum() / len(correct)
    return acc
```

```python
def train(model, dataloader, optimizer, criterion):
    
    model.train()
    
    epoch_loss = 0
    epoch_acc = 0

    for xb, yb, lengths in dataloader:
        xb, yb, lengths = xb.to(device), yb.to(device), lengths.to(device)
        optimizer.zero_grad()
        logits = model(xb, lengths)
        loss = criterion(logits, yb)
        acc = binary_accuracy(logits, yb)
        loss.backward()
        optimizer.step()

        epoch_loss += loss.item()
        epoch_acc += acc.item()

    return epoch_loss / len(dataloader), epoch_acc / len(dataloader)
```

```python
def validate(model, dataloader, criterion):

    model.eval()

    val_epoch_loss = 0
    val_epoch_acc = 0

    for xb, yb, lengths in dataloader:
        xb, yb, lengths = xb.to(device), yb.to(device), lengths.to(device)
        with torch.no_grad():
            logits = model(xb, lengths)
            loss = criterion(logits, yb)
        acc = binary_accuracy(logits, yb)
        val_epoch_loss += loss.item()
        val_epoch_acc += acc.item()  

    return val_epoch_loss/len(dataloader), val_epoch_acc/len(dataloader)
```

```python
def evaluate_model(model, dataloader):

    model.eval()
    
    all_logits = []
    all_labels = []

    for xb, yb, lengths in tqdm(dataloader):
        xb, yb, lengths = xb.to(device), yb.to(device), lengths.to(device)
        with torch.no_grad():
            logits = model(xb, lengths)
        all_logits += logits.tolist()
        all_labels += yb.tolist()
            
    all_logits = torch.tensor(all_logits)
    all_labels = torch.tensor(all_labels)

    y_pred_binary = torch.round(torch.sigmoid(all_logits)).cpu().numpy()
    y_true = all_labels.cpu().numpy()

    accuracy = accuracy_score(y_true, y_pred_binary)
    f1 = f1_score(y_true, y_pred_binary)
    roc_auc = roc_auc_score(y_true, all_logits.cpu().numpy())

    print("Accuracy:", accuracy)
    print("F1 Score:", f1)
    print("ROC AUC Score:", roc_auc)

    fpr, tpr, _ = roc_curve(y_true, all_logits.cpu().numpy())
    roc_auc = auc(fpr, tpr)

    plt.figure()
    plt.plot(fpr, tpr, color='darkorange', lw=2, label='ROC curve (area = %0.2f)' % roc_auc)
    plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('Receiver Operating Characteristic')
    plt.legend(loc="lower right")
    plt.show()
```

#### 5.1 Recurrent Neural Network

```python
class ModelRNN(nn.Module):
    def __init__(self, vocab_size, embedding_dim, hidden_dim, output_dim, n_layers, dropout, weights=None):
        super().__init__()
        if weights is not None:
            self.embedding = nn.Embedding.from_pretrained(weights, freeze=False,
                                                           padding_idx=0)
        else:
            self.embedding = nn.Embedding(vocab_size, embedding_dim)
        self.rnn = nn.RNN(embedding_dim, hidden_dim, n_layers, dropout = dropout)
        self.fc = nn.Linear(hidden_dim, output_dim)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x, lengths):
        # x: [batch, seq_len], lengths: [batch]
        emb = self.embedding(x)  # [batch, seq_len, embed_dim]

        # Упаковка — модель не будет обрабатывать паддинг
        packed = pack_padded_sequence(
            emb, lengths.cpu(), batch_first=True, enforce_sorted=False
        )
        packed_out, h_n = self.rnn(packed)
        # h_n: [num_layers, batch, hidden_dim]

        h_last = h_n[-1]  # [batch, hidden_dim] — последнее скрытое состояние
        h_last = self.dropout(h_last)
        logits = self.fc(h_last)  # [batch, num_classes]
        return logits
```

```python
model_rnn = ModelRNN(
    vocab_size=vocab_size, 
    embedding_dim=EMBEDDING_DIM, 
    hidden_dim=HIDDEN_DIM, 
    output_dim=OUTPUT_DIM, 
    n_layers=N_LAYERS, 
    dropout=DROPOUT,
    weights=full_weights
)
```

```python
model_rnn = model_rnn.to(device)
optimizer = optim.Adam(model_rnn.parameters())
```

```python
%%time
epoch_losses_rnn = {}
epoch_accuracies_rnn = {}
val_losses_rnn = {}
val_accuracies_rnn = {}

for epoch in tqdm(range(N_EPOCHS)):
  
  if (epoch + 1)%VAL_STEPS == 0:
    epoch_loss, epoch_accuracy = train(model_rnn, train_dataloader, optimizer, criterion)
    val_loss, val_accuracy = validate(model_rnn, val_dataloader, criterion)
    print(f'Epoch {epoch+1:02}, Epoch loss = {epoch_loss:.2f}, Epoch accuracy = {epoch_accuracy:.3f},',
          f'Validation loss = {val_loss:.2f}, Validation accuracy = {val_accuracy:.3f}')
    epoch_losses_rnn[epoch + 1] = epoch_loss
    epoch_accuracies_rnn[epoch + 1] = epoch_accuracy
    val_losses_rnn[epoch + 1] = val_loss
    val_accuracies_rnn[epoch + 1] = val_accuracy
  else:
    epoch_loss, epoch_accuracy = train(model_rnn, train_dataloader, optimizer, criterion)
    print(f'Epoch {epoch+1:02}, Epoch loss = {epoch_loss:.2f}, Epoch accuracy = {epoch_accuracy:.3f}')
    epoch_losses_rnn[epoch + 1] = epoch_loss
    epoch_accuracies_rnn[epoch + 1] = epoch_accuracy
```

```python
plt.figure(figsize=(12, 5))

#Loss
plt.subplot(1, 2, 1)
plt.plot(epoch_losses_rnn.keys(), epoch_losses_rnn.values(), label='Training Loss')
plt.plot(val_losses_rnn.keys(), val_losses_rnn.values(), label='Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

#Accuracy
plt.subplot(1, 2, 2)
plt.plot(epoch_accuracies_rnn.keys(), epoch_accuracies_rnn.values(), label='Training Accuracy')
plt.plot(val_accuracies_rnn.keys(), val_accuracies_rnn.values(), label='Validation Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

plt.show()
```

```python
evaluate_model(model_rnn, test_dataloader)
```

#### 5.2 Long Short‑Term Memory

```python
class ModelLSTM(nn.Module):
    def __init__(self, vocab_size, embedding_dim, hidden_dim, output_dim, n_layers, dropout, weights=None):
        super().__init__()
        if weights is not None:
            self.embedding = nn.Embedding.from_pretrained(weights, freeze=False,
                                                           padding_idx=0)
        else:
            self.embedding = nn.Embedding(vocab_size, embedding_dim)
        self.lstm = nn.LSTM(embedding_dim, hidden_dim, n_layers, dropout = dropout)
        self.fc = nn.Linear(hidden_dim, output_dim)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x, lengths):
        # x: [batch, seq_len], lengths: [batch]
        emb = self.embedding(x)  # [batch, seq_len, embed_dim]

        # Упаковка — модель не будет обрабатывать паддинг
        packed = pack_padded_sequence(
            emb, lengths.cpu(), batch_first=True, enforce_sorted=False
        )
        packed_out, (h_n, c_n) = self.lstm(packed)
        # h_n: [num_layers, batch, hidden_dim]

        h_last = h_n[-1]  # [batch, hidden_dim] — последнее скрытое состояние
        h_last = self.dropout(h_last)
        logits = self.fc(h_last)  # [batch, num_classes]
        return logits
```

```python
model_lstm = ModelLSTM(
    vocab_size=vocab_size, 
    embedding_dim=EMBEDDING_DIM, 
    hidden_dim=HIDDEN_DIM, 
    output_dim=OUTPUT_DIM, 
    n_layers=N_LAYERS, 
    dropout=DROPOUT,
    weights=full_weights
)
```

```python
model_lstm = model_lstm.to(device)
optimizer = optim.Adam(model_lstm.parameters())
```

```python
%%time
epoch_losses_lstm = {}
epoch_accuracies_lstm = {}
val_losses_lstm = {}
val_accuracies_lstm = {}

for epoch in tqdm(range(N_EPOCHS)):
  
  if (epoch + 1)%VAL_STEPS == 0:
    epoch_loss, epoch_accuracy = train(model_lstm, train_dataloader, optimizer, criterion)
    val_loss, val_accuracy = validate(model_lstm, val_dataloader, criterion)
    print(f'Epoch {epoch+1:02}, Epoch loss = {epoch_loss:.2f}, Epoch accuracy = {epoch_accuracy:.3f},',
          f'Validation loss = {val_loss:.2f}, Validation accuracy = {val_accuracy:.3f}')
    epoch_losses_lstm[epoch + 1] = epoch_loss
    epoch_accuracies_lstm[epoch + 1] = epoch_accuracy
    val_losses_lstm[epoch + 1] = val_loss
    val_accuracies_lstm[epoch + 1] = val_accuracy
  else:
    epoch_loss, epoch_accuracy = train(model_lstm, train_dataloader, optimizer, criterion)
    print(f'Epoch {epoch+1:02}, Epoch loss = {epoch_loss:.2f}, Epoch accuracy = {epoch_accuracy:.3f}')
    epoch_losses_lstm[epoch + 1] = epoch_loss
    epoch_accuracies_lstm[epoch + 1] = epoch_accuracy
```

#### 5.3 Gated Recurrent Unit

```python
class ModelGRU(nn.Module):
    def __init__(self, vocab_size, embedding_dim, hidden_dim, output_dim, n_layers, dropout, weights=None):
        super().__init__()
        if weights is not None:
            self.embedding = nn.Embedding.from_pretrained(weights, freeze=False,
                                                           padding_idx=0)
        else:
            self.embedding = nn.Embedding(vocab_size, embedding_dim)
        self.gru = nn.GRU(embedding_dim, hidden_dim, n_layers, dropout = dropout)
        self.fc = nn.Linear(hidden_dim, output_dim)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x, lengths):
        # x: [batch, seq_len], lengths: [batch]
        emb = self.embedding(x)  # [batch, seq_len, embed_dim]

        # Упаковка — модель не будет обрабатывать паддинг
        packed = pack_padded_sequence(
            emb, lengths.cpu(), batch_first=True, enforce_sorted=False
        )
        packed_out, h_n = self.gru(packed)
        # h_n: [num_layers, batch, hidden_dim]

        h_last = h_n[-1]  # [batch, hidden_dim] — последнее скрытое состояние
        h_last = self.dropout(h_last)
        logits = self.fc(h_last)  # [batch, num_classes]
        return logits
```

```python
model_gru = ModelGRU(
    vocab_size=vocab_size, 
    embedding_dim=EMBEDDING_DIM, 
    hidden_dim=HIDDEN_DIM, 
    output_dim=OUTPUT_DIM, 
    n_layers=N_LAYERS, 
    dropout=DROPOUT,
    weights=full_weights
)
```

```python
model_gru = model_gru.to(device)
optimizer = optim.Adam(model_gru.parameters())
```

```python
%%time
epoch_losses_gru = {}
epoch_accuracies_gru = {}
val_losses_gru = {}
val_accuracies_gru = {}

for epoch in tqdm(range(N_EPOCHS)):
  
  if (epoch + 1)%VAL_STEPS == 0:
    epoch_loss, epoch_accuracy = train(model_gru, train_dataloader, optimizer, criterion)
    val_loss, val_accuracy = validate(model_gru, val_dataloader, criterion)
    print(f'Epoch {epoch+1:02}, Epoch loss = {epoch_loss:.2f}, Epoch accuracy = {epoch_accuracy:.3f},',
          f'Validation loss = {val_loss:.2f}, Validation accuracy = {val_accuracy:.3f}')
    epoch_losses_gru[epoch + 1] = epoch_loss
    epoch_accuracies_gru[epoch + 1] = epoch_accuracy
    val_losses_gru[epoch + 1] = val_loss
    val_accuracies_gru[epoch + 1] = val_accuracy
  else:
    epoch_loss, epoch_accuracy = train(model_gru, train_dataloader, optimizer, criterion)
    print(f'Epoch {epoch+1:02}, Epoch loss = {epoch_loss:.2f}, Epoch accuracy = {epoch_accuracy:.3f}')
    epoch_losses_gru[epoch + 1] = epoch_loss
    epoch_accuracies_gru[epoch + 1] = epoch_accuracy
```

```python
plt.figure(figsize=(12, 5))

#Loss
plt.subplot(1, 2, 1)
plt.plot(epoch_losses_gru.keys(), epoch_losses_gru.values(), label='Training Loss')
plt.plot(val_losses_gru.keys(), val_losses_gru.values(), label='Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

#Accuracy
plt.subplot(1, 2, 2)
plt.plot(epoch_accuracies_gru.keys(), epoch_accuracies_gru.values(), label='Training Accuracy')
plt.plot(val_accuracies_gru.keys(), val_accuracies_gru.values(), label='Validation Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

plt.show()
```
